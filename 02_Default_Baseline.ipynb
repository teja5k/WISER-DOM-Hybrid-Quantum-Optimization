```python
import pandas as pd

# Load datasets
orders = pd.read_csv("input_order data.csv")
shipping = pd.read_csv("input_shipping_cost_data.csv")
inventory = pd.read_csv("input_capacity_planning.csv")
dock = pd.read_csv("input_dock_capacity.csv")
throughput = pd.read_csv("input_throughput_capacity.csv")

# -----------------------------
# Keep only required columns
# -----------------------------

orders_clean = orders[
    [
        "Plant",
        "MaterialNumber",
        "RequestedDeliveryDate",
        "OrderedQty_converted",
        "OrderedWeight",
        "OrderedVolume",
        "Order_SKU_Revenue",
        "Penaltyforpotentialcuts",
        "ZipCode",
        "LoadNumber",
    ]
].copy()

inventory_clean = inventory[
    [
        "LocationID",
        "MaterialID",
        "DATE",
        "Available_inventory",
    ]
].copy()

shipping_clean = shipping[
    [
        "Plant",
        "TargetZip",
        "Shipping_Cost",
        "Distance",
    ]
].copy()

dock_clean = dock[
    [
        "Plant",
        "Date",
        "Dock_Remaining",
    ]
].copy()

throughput_clean = throughput[
    [
        "Plant",
        "transportationplanningdate",
        "order_count",
    ]
].copy()

print("="*60)
print("PREPROCESSED DATA")
print("="*60)

print("Orders      :", orders_clean.shape)
print("Inventory   :", inventory_clean.shape)
print("Shipping    :", shipping_clean.shape)
print("Dock        :", dock_clean.shape)
print("Throughput  :", throughput_clean.shape)

print("\nSample Orders")
display(orders_clean.head())

print("\nSample Inventory")
display(inventory_clean.head())

print("\nSample Shipping")
display(shipping_clean.head())
```

```python
import pandas as pd

# Orders
orders['RequestedDeliveryDate'] = pd.to_datetime(
    orders['RequestedDeliveryDate'],
    errors='coerce'
)

# Inventory
inventory['DATE'] = pd.to_datetime(
    inventory['DATE'],
    errors='coerce'
)

# Dock
dock['Date'] = pd.to_datetime(
    dock['Date'],
    errors='coerce'
)

# Throughput
throughput['transportationplanningdate'] = pd.to_datetime(
    throughput['transportationplanningdate'],
    errors='coerce'
)

print("Dates converted successfully.")
```

```python
print("="*60)
print("Creating Lookup Dictionaries...")
print("="*60)

# Inventory Lookup
inventory_lookup = (
    inventory
    .groupby(['LocationID','MaterialID'])['Available_inventory']
    .max()
    .to_dict()
)

# Shipping Lookup
shipping_lookup = (
    shipping
    .set_index(['Plant','TargetZip'])
    [['Shipping_Cost','Distance']]
    .to_dict('index')
)

# Dock Lookup
dock_lookup = (
    dock
    .groupby('Plant')['Dock_Remaining']
    .mean()
    .to_dict()
)

# Throughput Lookup
throughput_lookup = (
    throughput
    .groupby('Plant')['util_pallets']
    .mean()
    .to_dict()
)

print("Inventory :", len(inventory_lookup))
print("Shipping  :", len(shipping_lookup))
print("Dock      :", len(dock_lookup))
print("Throughput:", len(throughput_lookup))
```

```python
candidate_plants = sorted(shipping["Plant"].unique())

print(candidate_plants)
print("\nTotal Candidate Plants:", len(candidate_plants))
```

```python
candidate_rows = []

for _, order in orders.iterrows():

    material = order["MaterialNumber"]
    zipcode = order["ZipCode"]

    for plant in candidate_plants:

        inv = inventory_lookup.get((plant, material), 0)

        ship = shipping_lookup.get((plant, zipcode), None)

        if ship is None:
            continue

        ship_cost = ship["Shipping_Cost"]
        distance = ship["Distance"]

        dock_remaining = dock_lookup.get(plant, 0)
        throughput_remaining = throughput_lookup.get(plant, 0)

        candidate_rows.append({

            "LoadNumber": order["LoadNumber"],

            "Plant": plant,

            "Material": material,

            "ZipCode": zipcode,

            "Revenue": order["Order_SKU_Revenue"],

            "Penalty": order["Penaltyforpotentialcuts"],

            "Inventory": inv,

            "ShippingCost": ship_cost,

            "Distance": distance,

            "DockRemaining": dock_remaining,

            "ThroughputRemaining": throughput_remaining

        })

candidate_df = pd.DataFrame(candidate_rows)

print(candidate_df.shape)

candidate_df.head()
```

```python
candidate_df["Inventory_OK"] = (
    candidate_df["Inventory"] > 0
)

candidate_df["Dock_OK"] = (
    candidate_df["DockRemaining"] > 0
)

candidate_df["Throughput_OK"] = (
    candidate_df["ThroughputRemaining"] > 0
)

candidate_df["Feasible"] = (
    candidate_df["Inventory_OK"] &
    candidate_df["Dock_OK"] &
    candidate_df["Throughput_OK"]
)
```

```python
print("="*60)
print("Candidate Assignment Summary")
print("="*60)

print("Total Candidate Rows :", len(candidate_df))

print("Feasible Assignments :", candidate_df["Feasible"].sum())

print("Unique Plants :", candidate_df["Plant"].nunique())

print("\nFeasible Percentage")

print(
    round(
        candidate_df["Feasible"].mean()*100,
        2
    ),
    "%"
)

print("\nAssignments per Plant")

print(
    candidate_df.groupby("Plant")["Feasible"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
print("="*60)
print("Creating Feasible Optimization Dataset")
print("="*60)

# Keep only feasible assignments
feasible_df = candidate_df[
    candidate_df["Feasible"] == True
].copy()

# Handle missing penalty values
feasible_df["Penalty"] = (
    feasible_df["Penalty"]
    .fillna(0)
)

print("Feasible Dataset Shape:")
print(feasible_df.shape)

print("\nColumns:")
print(feasible_df.columns.tolist())

print("\nSample Data:")
display(feasible_df.head())
```

```python
print("="*60)
print("Creating Adaptive Priority Score")
print("="*60)

from sklearn.preprocessing import MinMaxScaler

# Features to normalize
score_features = [
    "Revenue",
    "ShippingCost",
    "Inventory",
    "DockRemaining",
    "ThroughputRemaining",
    "Penalty"
]

# Create scaler
scaler = MinMaxScaler()

# Normalize values
normalized_values = scaler.fit_transform(
    feasible_df[score_features]
)

# Add normalized columns
for i, col in enumerate(score_features):
    feasible_df[col + "_Norm"] = normalized_values[:, i]


print("Normalization completed")

print("\nNew Columns:")
print(
    [col for col in feasible_df.columns if "_Norm" in col]
)

display(
    feasible_df.head()
)
```

```python
print("="*60)
print("Calculating Adaptive Priority Score")
print("="*60)


feasible_df["AdaptiveScore"] = (

    0.35 * feasible_df["Revenue_Norm"]

    - 0.25 * feasible_df["ShippingCost_Norm"]

    + 0.15 * feasible_df["Inventory_Norm"]

    + 0.10 * feasible_df["DockRemaining_Norm"]

    + 0.10 * feasible_df["ThroughputRemaining_Norm"]

    - 0.05 * feasible_df["Penalty_Norm"]

)


print("Adaptive Score Created")

print("\nScore Statistics:")
print(
    feasible_df["AdaptiveScore"].describe()
)
```

```python
print("="*60)
print("Adaptive Score Ranking Validation")
print("="*60)

# Pick one order
sample_order = feasible_df["LoadNumber"].iloc[0]

print("Sample LoadNumber:", sample_order)

display(
    feasible_df[
        feasible_df["LoadNumber"] == sample_order
    ]
    [
        [
            "LoadNumber",
            "Plant",
            "Revenue",
            "ShippingCost",
            "Inventory",
            "DockRemaining",
            "ThroughputRemaining",
            "AdaptiveScore"
        ]
    ]
    .sort_values(
        "AdaptiveScore",
        ascending=False
    )
)
```

```python
print("="*60)
print("Building Sparse QUBO Matrix")
print("="*60)


Q = {}

# Constraint penalty weight
A = 2.0


# ==================================
# Objective Function
# Maximize Adaptive Score
# ==================================

for _, row in qubo_input_df.iterrows():

    var = row["QubitVariable"]

    score = row["AdaptiveScore"]

    # Convert maximization to minimization
    Q[(var,var)] = -score



# ==================================
# Assignment Constraint
# One plant per Order-Material
# ==================================

for (load, material), group in qubo_input_df.groupby(
        ["LoadNumber","Material"]):

    variables = group["QubitVariable"].tolist()


    # Linear penalty
    for v in variables:

        if (v,v) in Q:
            Q[(v,v)] -= A
        else:
            Q[(v,v)] = -A



    # Quadratic penalty
    for i in range(len(variables)):

        for j in range(i+1,len(variables)):

            pair = (
                variables[i],
                variables[j]
            )

            Q[pair] = A



print("\nQUBO Construction Completed")


print("\nTotal QUBO Terms:")
print(len(Q))


print("\nSample Terms:")

for key,value in list(Q.items())[:10]:
    print(key,":",value)
```

```python
print("="*60)
print("Creating QUBO Energy Function")
print("="*60)


def calculate_qubo_energy(selected_variables, Q):

    energy = 0

    selected = set(selected_variables)

    for (i,j), coefficient in Q.items():

        if i in selected and j in selected:

            energy += coefficient

    return energy



print("Energy Function Created")
```

```python
from qiskit_optimization import QuadraticProgram


selected_vars_20 = list(mapping_20.values())

adaptive_20_qp = QuadraticProgram(
    name="Adaptive_DOM_QUBO_20"
)


# ----------------------------
# 1. Add binary variables
# ----------------------------

for i in range(len(selected_vars_20)):
    adaptive_20_qp.binary_var(name=f"x{i}")


# ----------------------------
# 2. Prepare linear coefficients
# ----------------------------

linear = {}

for new_i, old_i in mapping_20.items():

    coeff = qp.objective.linear.to_dict().get(old_i, 0)

    if coeff != 0:
        linear[new_i] = coeff



# ----------------------------
# 3. Prepare quadratic coefficients
# ----------------------------

quadratic = {}

reverse_mapping = {
    old:new
    for new,old in mapping_20.items()
}


for (old_i, old_j), coeff in qp.objective.quadratic.to_dict().items():

    if old_i in reverse_mapping and old_j in reverse_mapping:

        new_i = reverse_mapping[old_i]
        new_j = reverse_mapping[old_j]

        quadratic[(new_i,new_j)] = coeff



# ----------------------------
# 4. Create objective
# ----------------------------

adaptive_20_qp.minimize(
    linear=linear,
    quadratic=quadratic
)


print(adaptive_20_qp)
```

```python
import itertools
import numpy as np


def evaluate_qubo(problem, x):

    value = problem.objective.constant

    # linear terms
    linear = problem.objective.linear.to_array()

    for i in range(len(x)):
        value += linear[i] * x[i]

    # quadratic terms
    quadratic = problem.objective.quadratic.to_array()

    for i in range(len(x)):
        for j in range(len(x)):
            value += quadratic[i][j] * x[i] * x[j]

    return value



num_vars = adaptive_20_qp.get_num_vars()

best_value = float("inf")
best_solution = None


print("Searching Classical Optimal Solution...")
print("Variables:", num_vars)


for bits in itertools.product([0,1], repeat=num_vars):

    x = np.array(bits)

    value = evaluate_qubo(
        adaptive_20_qp,
        x
    )

    if value < best_value:
        best_value = value
        best_solution = x


print("\nClassical Solution:")
print(best_solution)

print("\nClassical Objective:")
print(best_value)
```
